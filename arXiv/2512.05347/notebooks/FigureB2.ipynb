```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np 
import os

from matplotlib.ticker import (MultipleLocator, AutoMinorLocator)
from matplotlib.lines import Line2D
import matplotlib.patches as mpatches
from PIL import Image


sns.set_context("talk")
```

```python
fig, axs = plt.subplots(ncols=3
                        ,nrows=2
                        ,figsize=(5*2,4*2)
                        ,constrained_layout = True
                        ,width_ratios=(3,1,3)
                       )

sns.set_context("talk")


labx=0.025
laby=1.025

temperature_map=[[50,300],[300,50]]

label_index=list("abcd")
for j, protocol in enumerate(["LH","HL"]):
    for i, temp in enumerate(temperature_map):
        temperature = temp[j]
        ax=axs[j,i*2]
        ax.text(labx, laby, '(%s)'%label_index[i+2*j],weight='bold' ,transform=ax.transAxes)
        ax.set_title("%d K"%temperature)
        img = np.asarray(Image.open('../figures/structures/md_%dK_%s.png'%(temperature,protocol)))
        ax.imshow(img)
        ax.set_yticks([])
        ax.set_xticks([])

    ax=axs[j,1]
    arrow = mpatches.FancyArrowPatch((0, 0.5), (1, 0.5),
                         mutation_scale=100,
                         transform=ax.transAxes)            
    ax.add_patch(arrow)
    ax.text(0.5, 0.7, ['Heating\nProtocol',"Cooling\nProtocol"][j] ,transform=ax.transAxes,ha='center')
    ax.axis('off')

plt.savefig("../figures/figB2.pdf", pad_inches=0.2,bbox_inches="tight")
```
