```python
from pymatgen.core import Structure
from pymatgen.io.lammps.data import LammpsData
from pymatgen.transformations.advanced_transformations import CubicSupercellTransformation
import numpy as np 
```

```python
structure=Structure.from_file("../data/structure/1526860.cif")
cubic_structure=CubicSupercellTransformation().apply_transformation(structure)

block_size=100
cubic_structure.make_supercell([block_size//cubic_structure.lattice.a,block_size//cubic_structure.lattice.b,block_size//cubic_structure.lattice.c])
cubic_structure_sorted=cubic_structure.get_sorted_structure()
LammpsData.from_structure(cubic_structure_sorted,atom_style="full").write_file("../result/structure/SiO2_alpha_quartz_block_%dnm.data"%(block_size//10))

atom_map={"Si":"SI1","O":"OM1"}
residue_map={"Si":"SI","O":"OM"}
result_str="Aplha quartz block\n"
result_str+="%d\n"%(len(cubic_structure_sorted.sites))
total_charge=0
composition_map={"Si":0,"O":0}
domain=np.array([[1e6,-1e6],[1e6,-1e6],[1e6,-1e6]])
bbox_padding=0
for i,site in enumerate(cubic_structure_sorted.sites):
    x,y,z=site.coords
    domain[0,0],domain[0,1]=min(x,domain[0,0]),max(x,domain[0,1])
    domain[1,0],domain[1,1]=min(y,domain[1,0]),max(y,domain[1,1])
    domain[2,0],domain[2,1]=min(z,domain[2,0]),max(z,domain[2,1])
bbox=np.diff(domain, axis=1)
for i,site in enumerate(cubic_structure_sorted.sites):
#    total_charge+=site.properties['charge']
    composition_map[site.species_string]+=1
    res_number=i+1
    x,y,z=site.coords
#    res_index=site.properties['atom_number']
    atom_name="%s"%(atom_map[site.species_string])
    res_name=residue_map[site.species_string]
    result_str+=f'{res_number:>5}{res_name:<5}{atom_name:>5}{i+1:>5}{(x-domain[0,0])/10+0.5*bbox_padding:8.4f}{(y-domain[1,0])/10+0.5*bbox_padding:8.4f}{(z-domain[2,0])/10+0.5*bbox_padding:>8.4f}\n'
result_str+="%f %f %f\n"%(bbox[0]/10+bbox_padding,bbox[1]/10+bbox_padding,bbox[2]/10+bbox_padding)
with open("../result/structure/SiO2_alpha_quartz_block_%dnm.gro"%(block_size//10),"w") as gro_file:
    gro_file.write(result_str)
print(composition_map)
```

```python
structure=LammpsData.from_file("../results/lammps/annealing/2nm/vashishta/results/final_structure.data").structure
structure=LammpsData.from_file("../results/structure/SiO2_alpha_quartz_block_6nm.data").structure

structure.add_site_property('index', np.arange(len(structure.sites)).tolist())
structure.add_site_property('velocities', np.arange(len(structure.sites)*3).reshape(-1,3))

lattice = structure.lattice
```

```python
radius=2

thickness=0.25
cx, cy, cz = lattice.a / 2, lattice.b / 2, lattice.c / 2

index_outside_boundary = {k: list() for k in ["Si","O"]}
index_inside = {k: list() for k in ["Si","O"]}
index_outside = {k: list() for k in ["Si","O"]}
for i, site in enumerate(structure.sites):
    x,y,z = site.coords
    rho=site.distance_from_point([cx, cy, z])
    if rho <= radius*10:
        index_inside[site.species_string].append(i)
    else:
        index_outside[site.species_string].append(i)
        if rho < (radius+thickness)*10 :
            index_outside_boundary[site.species_string].append(i)
```

```python
print(len(index_inside["O"]))
attach_hydrogen_list=list()
for site_index in index_outside_boundary["Si"]:
    Si_site=structure.sites[site_index]
    for i, nb_site in enumerate(structure.get_neighbors(Si_site,4)):
        nb_site_index=nb_site.properties["index"]
        if nb_site_index in index_inside["O"]:
            attach_hydrogen_list.append(nb_site_index)
            index_inside["O"].remove(nb_site_index)
print(len(index_inside["O"]))
```

```python
for site_index in attach_hydrogen_list:
    if site_index in index_outside:
        print(site_index)
    site=structure.sites[site_index]
    x,y,z = site.coords
    rho=site.distance_from_point([cx, cy, z])
    ex,ey=(x-cx)/rho,(y-cy)/rho    
    structure.append("H",(x-0.9584*ex,y-0.9584*ey,z),coords_are_cartesian=True,properties={'charge':0, 'velocities':np.array([0,0,0]), 'index':-1},validate_proximity=True)

structure.remove_sites(np.concatenate(list(index_inside.values())))
```
